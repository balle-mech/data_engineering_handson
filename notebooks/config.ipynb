**以下、All Runせず、一つ目のPythonセルを実行し、ウィジェットにカタログ・スキーマ名を入力してから、2つ目以降のセルを実行してください。**

```python
# カタログ・スキーマ名を設定するウィジェット機能を作成
dbutils.widgets.text("catalog_name", "my_lab")
dbutils.widgets.text("schema_name", "handson")
```

```python
# ウィジェットからカタログ・スキーマ名を取得
catalog_name = dbutils.widgets.get("catalog_name")
schema_name = dbutils.widgets.get("schema_name")

# auditテーブル名を設定
bronze_audit_table_name = "bronze_audit"
bronze_audit_table_path = f"{catalog_name}.{schema_name}.{bronze_audit_table_name}"
silver_audit_table_name = "silver_audit"
silver_audit_table_path = f"{catalog_name}.{schema_name}.{silver_audit_table_name}"

# usageテーブル名を設定
bronze_usage_table_name = "bronze_usage"
bronze_usage_table_path = f"{catalog_name}.{schema_name}.{bronze_usage_table_name}"
silver_usage_table_name = "silver_usage"
silver_usage_table_path = f"{catalog_name}.{schema_name}.{silver_usage_table_name}"

# ソースファイルを配置する Volume 名を設定
source_volume_name = "unit_mtg_handson"
audit_csv_path = f"/Volumes/{catalog_name}/{schema_name}/{source_volume_name}/audit"
usage_csv_path = f"/Volumes/{catalog_name}/{schema_name}/{source_volume_name}/usage"

# チェックポイントボリューム（AutoLoaderで使用）
checkpoint_volume_name = "checkpoint_volume_01"
checkpoint_volume_path = f"/Volumes/{catalog_name}/{schema_name}/{checkpoint_volume_name}"
```

```python
print("catalog_name: " + catalog_name)
print("schema_name: " + schema_name)

print("bronze_audit_table_path: " + bronze_audit_table_path)
print("silver_audit_table_path:" + silver_audit_table_path)

print("bronze__usage_table_path: " + bronze_usage_table_path)
print("silver_usage_table_path:" + silver_usage_table_path)

print("audit_csv_path: " + audit_csv_path)
print("usage_csv_path: " + usage_csv_path)
print("checkpoint_volume_path: " + checkpoint_volume_path)
```
